```python
import tensorflow as tf
import numpy as np
```

```python
X = np.array([1, 2, 3, 4, 5, 6, 7, 8])
y = np.array([11, 12, 13, 14, 15, 16, 17, 18])

X = tf.constant(X, dtype=tf.float32)
y = tf.constant(y, dtype=tf.float32)

X, y
```

# Steps for modelling in Tensorflow
* **Creating a model**: Define input, Output and Hidden layers of a deep learning model.
* **Compiling a model**: Define the loss function (eg. MSE, MAE, Hybrid) and optimizers (eg. Gradient Descent) and evaluation metrics.
* **Fitting a model**: Letting the model find out patterns between X and y.

```python
# set random seed
tf.random.set_seed(42)

# Create model using Sequential API

model = tf.keras.Sequential([
    tf.keras.layers.Dense(1)
])

# Compile the model

model.compile(loss = tf.keras.losses.mae, #mae = Mean absolute error
              optimizer = tf.keras.optimizers.SGD(), #sgd stands for stochasitc gradient descent
              metrics = ["mae"]
              )

# Fit the model
X = tf.reshape(X, [-1, 1])
model.fit(X, y, epochs=100)
```

```python
model.predict(tf.constant([9.0]))
```

# Improving our model
1. **Making Changes in Making the model**: Adding the number of layers or adding the number of neurons in each layer also adding the activation function(relu, tanh, etc).

2. **Making changes in the compilation step**: Changing the optimizer, adding the **learning rate** parameter to improve models learning rate at each iteration.

3. **Making Changes in the fitting step**: Increasing the number of epochs(iterations) to make model find patterns between the data more effficiently.

```python
# Carriying out 1st and 2nd step

# set random seed
tf.random.set_seed(42)

# Create model using Sequential API

model = tf.keras.Sequential([
    tf.keras.layers.Dense(1),
    tf.keras.layers.Dense(1),
    tf.keras.layers.Dense(1)
])

# Compile the model

model.compile(loss = tf.keras.losses.mae, #mae = Mean absolute error
              optimizer = tf.keras.optimizers.Adam(learning_rate = 0.01), #sgd stands for stochasitc gradient descent
              metrics = ["mae"]
              )

# Fit the model
X = tf.reshape(X, [-1, 1])
model.fit(X, y, epochs=100)
```

```python
model.predict(tf.constant([9.0]))
```

# Creating new dataset and dividing iy into three sets.
1. **Train Dataset:**A section of data on which model is to be trained. Covers mostly 70-80% of the main dataset.
2. **Validation Set:**A  section of data on which model is tuned. Covers mostly 10-15% of the main dataset.
3. **Test Dataset:** A section of data which is used to check the accuracy of the model. Covers mostly 15-20% of the main dataset.
